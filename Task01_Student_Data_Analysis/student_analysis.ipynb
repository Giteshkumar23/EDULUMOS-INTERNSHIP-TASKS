```python

import pandas as pd

df = pd.read_csv("StudentsPerformance.csv")

print("Dataset shape:", df.shape)

display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

```

```python

print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nStatistical summary:")
display(df.describe())
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

score_columns = [
    "math score",
    "reading score",
    "writing score"
]

df[score_columns].hist(
    figsize=(12, 4),
    bins=15,
    edgecolor="black"
)

plt.suptitle("Distribution of Student Scores")
plt.tight_layout()
plt.show()
```

```python

preparation_scores = df.groupby("test preparation course")[
    ["math score", "reading score", "writing score"]
].mean()

display(preparation_scores.round(2))

preparation_scores.plot(
    kind="bar",
    figsize=(9, 5),
    edgecolor="black"
)

plt.title("Average Scores by Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python

gender_scores = df.groupby("gender")[
    ["math score", "reading score", "writing score"]
].mean()

display(gender_scores.round(2))

gender_scores.plot(
    kind="bar",
    figsize=(8, 5),
    edgecolor="black"
)

plt.title("Average Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python

df["average_score"] = df[
    ["math score", "reading score", "writing score"]
].mean(axis=1)

print("Average score calculated successfully.")

display(
    df[["math score", "reading score", "writing score", "average_score"]].head()
)

print("\nOverall average score:", round(df["average_score"].mean(), 2))
```

```python

parent_education_scores = df.groupby(
    "parental level of education"
)["average_score"].mean().sort_values(ascending=False)

display(parent_education_scores.round(2))

parent_education_scores.plot(
    kind="bar",
    figsize=(10, 5),
    edgecolor="black"
)

plt.title("Average Student Score by Parental Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Average Student Score")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="average_score",
    bins=20,
    kde=True
)

plt.title("Distribution of Average Student Scores")
plt.xlabel("Average Score")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

correlation = df[
    ["math score", "reading score", "writing score", "average_score"]
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Student Scores")
plt.tight_layout()
plt.show()
```
