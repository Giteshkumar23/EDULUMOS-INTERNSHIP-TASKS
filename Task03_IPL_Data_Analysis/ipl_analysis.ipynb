```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path.cwd() / "data"

matches = pd.read_csv(
    DATA_DIR / "matches_updated_ipl_upto_2025.csv"
)

print("Dataset shape:", matches.shape)
print("\nSeasons available:")
print(sorted(matches["season"].dropna().astype(str).unique()))

print("\nLatest seasons:")
print(matches["season"].value_counts().head(10))

print("\nMissing values:")
display(matches.isnull().sum().sort_values(ascending=False).head(10))

display(matches.head())
```

```python
matches["date"] = pd.to_datetime(matches["date"], errors="coerce")

matches["season"] = matches["season"].astype(str)

print("Total matches:", len(matches))
print("Total seasons:", matches["season"].nunique())
print("First season:", matches["season"].min())
print("Latest season:", matches["season"].max())

season_matches = matches["season"].value_counts().sort_index()

display(season_matches.rename("Matches").reset_index())
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=season_matches.index,
    y=season_matches.values
)

plt.title("Number of IPL Matches by Season")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "matches_by_season.png", dpi=150)
plt.show()
```

```python
ipl_2026 = pd.read_csv(DATA_DIR / "ipl_2026_ball_by_ball.csv")

print("Dataset shape:", ipl_2026.shape)
print("\nSeason values:")
print(ipl_2026["season"].value_counts(dropna=False))

print("\nMissing values:")
display(ipl_2026.isnull().sum().sort_values(ascending=False))

print("\nLast 10 rows:")
display(ipl_2026.tail(10))
```

```python
ipl_2026["runs_off_bat"] = pd.to_numeric(
    ipl_2026["runs_off_bat"], errors="coerce"
)

team_runs = (
    ipl_2026.groupby("batting_team")["runs_off_bat"]
    .sum()
    .sort_values(ascending=False)
)

display(team_runs.reset_index(name="Runs Scored"))

plt.figure(figsize=(10, 6))
team_runs.sort_values().plot(kind="barh")

plt.title("Total Runs Scored by Team - IPL 2026")
plt.xlabel("Runs Scored")
plt.ylabel("Team")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "team_runs_2026.png", dpi=150)
plt.show()
```

```python
matches_2025 = pd.read_csv(
    DATA_DIR / "matches_updated_ipl_upto_2025.csv"
)

deliveries_2025 = pd.read_csv(
    DATA_DIR / "deliveries_updated_ipl_upto_2025.csv"
)

matches_2025["season"] = matches_2025["season"].astype(str)

matches_2025 = matches_2025[
    matches_2025["season"] == "2025"
].copy()

print("2025 matches:", len(matches_2025))
print("Match dataset shape:", matches_2025.shape)
print("Delivery dataset shape:", deliveries_2025.shape)

display(matches_2025.head())
display(deliveries_2025.head())
```

```python
team_wins_2025 = (
    matches_2025["winner"]
    .dropna()
    .value_counts()
)

display(team_wins_2025.reset_index(name="Matches Won"))

plt.figure(figsize=(10, 6))
team_wins_2025.sort_values().plot(kind="barh")

plt.title("Matches Won by Each Team - IPL 2025")
plt.xlabel("Matches Won")
plt.ylabel("Team")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "team_wins_2025.png", dpi=150)
plt.show()
```

```python
deliveries_2025["batsman_runs"] = pd.to_numeric(
    deliveries_2025["batsman_runs"], errors="coerce"
)

top_batters_2025 = (
    deliveries_2025.groupby("batsman")["batsman_runs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(top_batters_2025.reset_index(name="Runs Scored"))

plt.figure(figsize=(10, 6))
top_batters_2025.sort_values().plot(kind="barh")

plt.title("Top 10 Run Scorers - IPL 2025")
plt.xlabel("Runs Scored")
plt.ylabel("Batter")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "top_batters_2025.png", dpi=150)
plt.show()
```

```python
non_bowler_dismissals = [
    "run out",
    "retired hurt",
    "retired out",
    "obstructing the field"
]

wickets_2025 = deliveries_2025[
    deliveries_2025["dismissal_kind"].notna()
    & ~deliveries_2025["dismissal_kind"].str.lower().isin(
        non_bowler_dismissals
    )
]

top_bowlers_2025 = (
    wickets_2025.groupby("bowler")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

display(top_bowlers_2025.reset_index(name="Wickets"))

plt.figure(figsize=(10, 6))
top_bowlers_2025.sort_values().plot(kind="barh")

plt.title("Top 10 Wicket-Takers - IPL 2025")
plt.xlabel("Wickets")
plt.ylabel("Bowler")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "top_bowlers_2025.png", dpi=150)
plt.show()
```
